```python
# import section 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# seccion de carga de datos
base_df= pd.read_csv('../../data/csv/history.csv') 
```

```python
# Crear una copia del dataset original
df = base_df.copy()

df.columns = df.columns.str.strip()

# Ejercicio 1. Filtrar meses donde temp_max > 35°C
filtro_temp_max = df[df['temp_max'] > 35]


# Mostrar resultados
print("Meses con temp_max > 35°C:")
print(filtro_temp_max[['year_month', 'temp_max']])

plt.figure(figsize=(10, 5))
plt.bar(filtro_temp_max['year_month'], filtro_temp_max['temp_max'], color='tomato')
plt.title('Meses con temp_max > 35°C')
plt.xlabel('Mes')
plt.ylabel('Temperatura Máxima (°C)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Crear una copia del dataset original
df = base_df.copy()

df.columns = df.columns.str.strip()

# Ejercicio 2. Filtrar meses donde viento_vel_max > 50 km/h
filtro_viento_max = df[df['viento_vel_max'] > 50]
```

```python
# Crear una copia del dataset original
df = base_df.copy()

df.columns = df.columns.str.strip()

# Ejercicio 3. Crear columna "temp_categoria"
def clasificar_temp(temp):
    if temp < 10:
        return 'Frío'
    elif 10 <= temp <= 25:
        return 'Templado'
    else:
        return 'Caluroso'

df['temp_categoria'] = df['temp_media'].apply(clasificar_temp)
```

```python
# Crear una copia del dataset original
df = base_df.copy()

df.columns = df.columns.str.strip()

# Ejercicio 4. Crear columna "estado_rocio"
def clasificar_rocio(temp_rocio):
    if temp_rocio < 10:
        return 'Seco'
    elif 10 <= temp_rocio <= 20:
        return 'Moderado'
    else:
        return 'Húmedo'

df['estado_rocio'] = df['temp_rocio'].apply(clasificar_rocio)
```

```python
# Crear una copia del dataset original
df = base_df.copy()

df.columns = df.columns.str.strip()

# Ejercicio 5. Filtrar registros donde pres_atm_nivel_mar < 1000 hPa o > 1030 hPa
filtro_presion = df[(df['pres_atm_nivel_mar'] < 1000) | (df['pres_atm_nivel_mar'] > 1030)]

# Mostrar resultados
print("\nRegistros con presión anómala (pres_atm_nivel_mar < 1000 o > 1030 hPa):")
print(filtro_presion[['year_month', 'pres_atm_nivel_mar']])

if not filtro_presion.empty:
    plt.figure(figsize=(10, 5))
    plt.bar(filtro_presion['year_month'], filtro_presion['pres_atm_nivel_mar'], color='royalblue')
    plt.title('Registros con Presión Anómala')
    plt.xlabel('Mes')
    plt.ylabel('Presión (hPa)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Crear una copia del dataset original
df = base_df.copy()

df.columns = df.columns.str.strip()

# Ejercicio 6. Filtrar donde humedad_relativa_media > 90% y la diferencia entre temp_rocio y temp_media sea <= 2°C
filtro_humedad = df[(df['humedad_relativa_media'] > 90) & (abs(df['temp_rocio'] - df['temp_media']) <= 2)]

# Mostrar resultados
print("\nRegistros con alta humedad y poca diferencia entre temp_media y temp_rocio:")
print(filtro_humedad[['year_month', 'humedad_relativa_media', 'temp_media', 'temp_rocio']])

if not filtro_humedad.empty:
    plt.figure(figsize=(10, 5))
    plt.bar(filtro_humedad['year_month'], filtro_humedad['humedad_relativa_media'], color='forestgreen')
    plt.title('Alta Humedad y Baja Diferencia de Temperatura')
    plt.xlabel('Mes')
    plt.ylabel('Humedad Relativa (%)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Crear una copia del dataset original
df = base_df.copy()

df.columns = df.columns.str.strip()

# Ejercicio 7. Agrupar por "year_month" y calcular la temperatura media promedio por mes
promedio_mensual = df.groupby('year_month')['temp_media'].mean().reset_index()

# Mostrar resultados
print("\nTemperatura media promedio por mes:")
print(promedio_mensual)

plt.figure(figsize=(10, 5))
plt.plot(promedio_mensual['year_month'], promedio_mensual['temp_media'], marker='o', linestyle='-', color='orange')
plt.title('Temperatura Media Promedio por Mes')
plt.xlabel('Mes')
plt.ylabel('Temperatura Media (°C)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Crear una copia del dataset original
df = base_df.copy()

df.columns = df.columns.str.strip()

# Ejercicio 11. Crear columna "relacion_precip_viento" 
df['relacion_precip_viento'] = df['precip_total'] / df['viento_vel_med']
```

```python
# Crear una copia del dataset original
df = base_df.copy()

# Limpiar nombres de columnas
df.columns = df.columns.str.strip()

# Manejar outliers (recortar valores extremos al percentil 5 y 95)
df['precip_total'] = df['precip_total'].clip(df['precip_total'].quantile(0.05), df['precip_total'].quantile(0.95))
df['temp_media'] = df['temp_media'].clip(df['temp_media'].quantile(0.05), df['temp_media'].quantile(0.95))
df['viento_vel_med'] = df['viento_vel_med'].clip(df['viento_vel_med'].quantile(0.05), df['viento_vel_med'].quantile(0.95))

# Normalizar las métricas (Min-Max Scaling)
df['precip_norm'] = (df['precip_total'] - df['precip_total'].min()) / (df['precip_total'].max() - df['precip_total'].min())
df['temp_norm'] = 1 - abs(df['temp_media'] - 20) / (df['temp_media'].max() - df['temp_media'].min())
df['viento_norm'] = (df['viento_vel_med'].max() - df['viento_vel_med']) / (df['viento_vel_med'].max() - df['viento_vel_med'].min())

# Calcular ponderaciones dinámicas (inverso de la desviación estándar)
pesos = 1 / df[['precip_norm', 'temp_norm', 'viento_norm']].std()
pesos /= pesos.sum()  # Normalizar para que la suma sea 1

# Calcular la "puntuación de construcción" con ponderación adaptativa
df['puntuacion_construccion'] = (
    df['precip_norm'] * pesos['precip_norm'] +
    df['temp_norm'] * pesos['temp_norm'] +
    df['viento_norm'] * pesos['viento_norm']
)

# Escalar la puntuación al rango [0, 100]
df['puntuacion_construccion'] = (df['puntuacion_construccion'] - df['puntuacion_construccion'].min()) / \
                                (df['puntuacion_construccion'].max() - df['puntuacion_construccion'].min()) * 100

# Extraer el año y el mes del campo 'year_month'
df['year'] = df['year_month'].str[:4].astype(int)
df['month'] = df['year_month'].str[5:7].astype(int)

# Encontrar el mejor mes de cada año
mejor_mes_por_ano = df.loc[df.groupby('year')['puntuacion_construccion'].idxmax()]

# Convertir el mes numérico en nombre del mes
mejor_mes_por_ano['mes_nombre'] = mejor_mes_por_ano['month'].apply(lambda x: 
    ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio', 'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre'][x-1])

# Mostrar los resultados
print("El mejor mes de cada año para construir:")
print(mejor_mes_por_ano[['year', 'mes_nombre', 'puntuacion_construccion']])

# Visualización de los mejores meses por año
plt.figure(figsize=(10, 6))
plt.bar(mejor_mes_por_ano['year'].astype(str), mejor_mes_por_ano['puntuacion_construccion'])
plt.title('Mejor mes de cada año para construir')
plt.xlabel('Año')
plt.ylabel('Puntuación de construcción')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
