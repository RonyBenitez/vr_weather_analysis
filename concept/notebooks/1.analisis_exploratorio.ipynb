```python
# import section 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# seccion de carga de datos
base_df= pd.read_csv('../../data/csv/history.csv') 
```

```python
# Crear una copia del dataset original
df = base_df.copy()

df.columns = df.columns.str.strip()

# Ejercicio 1. Filtrar meses donde temp_max > 35°C
filtro_temp_max = df[df['temp_max'] > 35]

# Ejercicio 2. Filtrar meses donde viento_vel_max > 50 km/h
filtro_viento_max = df[df['viento_vel_max'] > 50]

# Ejercicio 3. Crear columna "temp_categoria"
def clasificar_temp(temp):
    if temp < 10:
        return 'Frío'
    elif 10 <= temp <= 25:
        return 'Templado'
    else:
        return 'Caluroso'

df['temp_categoria'] = df['temp_media'].apply(clasificar_temp)

# Ejercicio 4. Crear columna "estado_rocio"
def clasificar_rocio(temp_rocio):
    if temp_rocio < 10:
        return 'Seco'
    elif 10 <= temp_rocio <= 20:
        return 'Moderado'
    else:
        return 'Húmedo'

df['estado_rocio'] = df['temp_rocio'].apply(clasificar_rocio)

# Ejercicio 5. Filtrar registros donde pres_atm_nivel_mar < 1000 hPa o > 1030 hPa
filtro_presion = df[(df['pres_atm_nivel_mar'] < 1000) | (df['pres_atm_nivel_mar'] > 1030)]

# Ejercicio 6. Filtrar donde humedad_relativa_media > 90% y la diferencia entre temp_rocio y temp_media sea <= 2°C
filtro_humedad = df[(df['humedad_relativa_media'] > 90) & (abs(df['temp_rocio'] - df['temp_media']) <= 2)]

# Ejercicio 7. Agrupar por "year_month" y calcular la temperatura media promedio por mes
promedio_mensual = df.groupby('year_month')['temp_media'].mean().reset_index()

# Ejercicio 8. Crear columna "amplitud_termica"
df['amplitud_termica'] = df['temp_max'] - df['temp_min']

# Ejercicio 9. Crear columna "indice_confort"
df['indice_confort'] = df['temp_media'] + (df['humedad_relativa_media'] / 100)

# Ejercicio 10. Crear columna "indice_calor" 
df['indice_calor'] = df['temp_media'] + (0.2 * df['humedad_relativa_media'])

# Ejercicio 11. Crear columna "relacion_precip_viento" 
df['relacion_precip_viento'] = df['precip_total'] / df['viento_vel_med']

# Ejercicio 12 (El ejercicio nuevo que se agrego luego). Filtrar meses ideales para la construcción
meses_ideales = df[(df['temp_media'] >= 10) & (df['temp_media'] <= 25) & 
                   (df['viento_vel_max'] <= 50) & 
                   (df['humedad_relativa_media'] <= 90)]

# Mostrar resultados
print("Meses con temp_max > 35°C:")
print(filtro_temp_max[['year_month', 'temp_max']])

print("\nMeses con viento_vel_max > 50 km/h:")
print(filtro_viento_max[['year_month', 'viento_vel_max']])

print("\nRegistros con presión anómala (pres_atm_nivel_mar < 1000 o > 1030 hPa):")
print(filtro_presion[['year_month', 'pres_atm_nivel_mar']])

print("\nRegistros con alta humedad y poca diferencia entre temp_media y temp_rocio:")
print(filtro_humedad[['year_month', 'humedad_relativa_media', 'temp_media', 'temp_rocio']])

print("\nTemperatura media promedio por mes:")
print(promedio_mensual)

print("\nMeses ideales para la construcción (temp_media entre 10°C y 25°C, viento <= 50 km/h, humedad <= 90%):")
print(meses_ideales[['year_month', 'temp_media', 'viento_vel_max', 'humedad_relativa_media']])

# Guardar el DataFrame transformado a un nuevo CSV (Recomendacion de GPT)
df.to_csv('../../data/csv/history_clasificado.csv', index=False)

print("\nTransformaciones completadas y guardadas en 'history_clasificado.csv'.")
```
